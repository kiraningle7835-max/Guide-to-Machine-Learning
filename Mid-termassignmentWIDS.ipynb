```python
#importing pandas and matplotlib modules
import pandas as pd
import matplotlib.pyplot as plt
#using read_csv func to open and read the file.The file will be stored in the form of a dataframe
df=pd.read_csv(r"\Users\Sai Ingle\Documents\student_bigdata.csv")
```

```python
#using df.head() to print the first 5 rows of the dataframe.
print(df.head())
```

```python
#using df.shape() to print the shape of the dataframe in the form of (no of rows,no of columns).
print(df.shape)
```

```python
#using df.columns to print the column names of the dataframe in the form of a list and showing the dtype as well at the end outside the list.
print(df.columns)
```

```python
#df.isna() returns a dataframe with boolean values as True(1) for missing values and False(0) for non missing values.Then we can use sum func to calculate the sum of all 1s thus finding the no of missing values in each column of the dataframe  
print(df.isna().sum())
```

```python
#Creating a new dataframe by dropping non numerical columns from the old dataframe.Then we can use mean(),median() and std() to find the mean,median and std deviation of each column of the dataframe. 
newdf=df.drop(["Student_ID","Gender","Final_Result"],axis=1)
print('Mean of values in each column:\n',newdf.mean(),'\nMedian of values in each column:\n',newdf.median(),'\nStd. deviation of values in each column:\n',newdf.std())
#We should note that this parameters are calculated while ignoring the missing values.For ex if we have to find the mean of the column which looks like this series - [1,2,3,4,5,None] then the mean would be (1+2+3+4+5)/5 which is 3.
```

```python
#importing the seaborn module to draw the histogram by using the histplot func and then using the plt.show() to show the graph on screen.
import seaborn as sns
sns.histplot(df["Math_Score"])
plt.title("Histogram showing Mathscores Vs Frequency distribution")
plt.xlabel("Math Scores")
plt.grid(axis="y",color="Black")
plt.show()
#we can also draw the histogram ny using matplotlib.
```

```python
#using the matplotlib to plot the histogram and plt.show() to show the graph on the screen.
plt.hist(df["Math_Score"],color="Red")
plt.title("Histogram showing Mathscores Vs Frequency distribution")
plt.xlabel("Math Scores")
plt.ylabel("Count")
plt.grid(axis="y",color="bLack")
plt.show()
```

```python
#using value_counts() which returns a series with each element like,here "Pass" and "Fail",as the index names and their frequency as the element of the series.
barinfo=df["Final_Result"].value_counts()
#using the bar() func to plot the bargraph and plt.show() to show the graph on screen.
plt.bar(barinfo.index,barinfo,color="Green")
plt.title("Bargraph showing no of students who have passed and those failed in the exam")
plt.xlabel("Pass or Fail")
plt.ylabel("Count of students")  
plt.show()
```

```python
#using the scatter() func to plot the scatter plot and plt.show() to show the graph on the screen.
plt.scatter(df["Study_Hours"],df["Math_Score"],c="indigo",alpha=0.9)
plt.xlabel("Study Hours")
plt.ylabel("Math Scores")
plt.title("Scatter plot showing distribution of Study-Hours of students Vs their math scores")
plt.show()
```
